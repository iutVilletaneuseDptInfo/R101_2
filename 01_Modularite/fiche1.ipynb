## Pourquoi découper son code en modules ?

Un programme qui grossit dans un seul fichier devient vite illisible, et impossible à partager sereinement à plusieurs. La modularité consiste à découper un programme en plusieurs fichiers (*modules*) regroupés de manière logique (*packages*), pour que le code reste lisible, réutilisable, et que plusieurs personnes puissent y travailler en parallèle.


Les modules et packages permettent de découper un programme en plusieurs fichiers et de regrouper ceux-ci de manière logique.

## Python en dehors des notebooks

**Python interactif :** la commande `python` ou `python3` dans un terminal transforme celui-ci en une console Python interactive. Chaque instruction saisie est directement exécutée. La commande `quite()` permet d'interrompre la session python. 

**Fichiers de code Python :** il est possible d'écrire du code Python dans un fichier (extension `.py`, encodage UTF-8) que l'on exécute dans un terminal avec la commande `python nom_du_fichier.py`.

## Modules

Un module est un fichier de code Python (nom de fichier = nom du module suffixé de `.py`) que l'on peut inclure dans un autre fichier Python ou un notebook, pour y appeler les fonctions qui y sont définies. Il faut au préalable **importer** le module. Il existe deux manières de le faire.

**`import ... as ...`** : les fonctions sont appelées via le nom donné au module, suivi d'un point puis du nom de la fonction.

```python
# Exemple (illustratif -- suppose l'existence d'un fichier module.py)
import module as nom_module

nom_module.fonction(val1, val2)  # fonction est définie dans module
````

**`from ... import ...`** : les fonctions sont appelées directement, sans préfixe.

```python
# Exemple (illustratif)
from module import fonction, autre_fonction  # etc

fonction(val1, val2)
````

**Remarque :** seules les fonctions dont les noms suivent le mot-clé `import` sont utilisables. Pour importer toutes les fonctions du module :

```python
from nom_du_module import * # déconseillé : on ne sait plus d'où viennent les fonctions, et des noms peuvent être écrasés
````

**Attention :** les fonctions ainsi importées écrasent les fonctions de même nom déjà définies dans le script.

### Utiliser des modules comme des scripts

Un module peut être utilisé dans d'autres scripts, mais aussi contenir du code qui ne s'exécute que lorsque ce module est le fichier exécuté directement (`python nom_du_module.py`) :

```python
if __name__ == '__main__':
    # Code qui sera exécuté uniquement si le module est exécuté
    # directement (pas inclus dans un autre script)
```

## 3. Packages en Python

Les packages permettent de regrouper logiquement des modules ensemble. Un package correspond à un répertoire : les fichiers (et sous-répertoires) à l'intérieur correspondent aux modules (et sous-packages) du package.

On importe un module d'un package selon l'une des deux méthodes usuelles :

```python
# Exemples (illustratifs)
import package.module as nom_module
from package.module import nom_fonction1, nom_fonction2  # etc
````

### Utiliser des modules et packages qui ne sont pas dans le répertoire courant

Si le module ou le package n'est pas dans le même répertoire que le script (ou notebook) qui les utilise, Python ne le trouve pas et cela engendre une erreur. Pour corriger ce problème, il faut ajouter le répertoire contenant le module (ou package) au chemin de recherche de Python, à l'aide du module `sys` :

```python
import sys

# chemin doit être une chaîne de caractères correspondant au chemin (relatif ou absolu)
# du répertoire contenant le module ou package que l'on souhaite importer
sys.path.append("chemin")
```

## 4. Tests unitaires avec pytest

Il est possible d'exécuter tous les tests unitaires d'un package ou d'un module à l'aide du programme `pytest`.

* La commande `pytest fichier.py` exécute toutes les fonctions commençant par `test_` définies dans `fichier.py`. Ces fonctions doivent être des fonctions de tests unitaires utilisant `assert` (selon le modèle vu en cours).
* La commande `pytest` (sans argument) considère tous les fichiers du répertoire courant (et de ses sous-répertoires) dont le nom commence par `test_`, et exécute toutes les fonctions `test_...` qui s'y trouvent.
