## Exercice 1 : Algorithme de César**

Vous voulez pouvoir échanger des messages codés avec vos camarades, sans que tout le monde puisse les lire directement : c'est l'occasion de manipuler un vrai algorithme de chiffrement, et de l'organiser proprement en module réutilisable.

Le chiffrement de César repose sur la rotation de l’alphabet d’un nombre fixé de caractères. Par exemple, si on considère une rotation de 5 caractères, on obtient la table de conversion ci-dessous :
* Caractère initial : a, b , c, d, e, ...
* Caractère transformé : f, g, h, i, j, ...

La fonction `chiffre_cesar(s,n)` définie ci-dessous chiffre le message `s` avec un décalage de `n` caractères. Elle transforme uniquement les lettres a, b, c, ..., z (minuscules ou majuscules) sans toucher aux signes de ponctuation ou autres caractères spéciaux.

**NB :** La fonction `ord(c)` retourne le code Unicode du caractère `c` et la fonction `chr(i)` retourne la lettre dont le code Unicode est `i`. Par exemple :

````python
unicode = ord("a")           # unicode contient 97 car le code Unicode 
                             # de "a" est 97
lettre = chr(98)             # lettre contient la valeur "b" car 
                             # son code Unicode est 98
print(chr(ord("a")) == "a")  # Affiche True
````

````python
def chiffre_lettre(c, n):
    """Retourne le chiffrement de la lettre c décalée de n caractères (si c n'est pas une lettre, elle est retournée inchangée).

    :param c: le caractère à chiffrer.
    :type c: str
    :param n: le décalage à appliquer.
    :returns: la lettre chiffrée, ou c inchangé si ce n'est pas une lettre."""
    if 'a' <= c and c <= 'z':  # si c est une lettre
        chiffree = chr((ord(c) + n - ord('a')) % 26 + ord('a'))  # décalée de n
    elif 'A' <= c and c <= 'Z':
        chiffree = chr((ord(c) + n - ord('A')) % 26 + ord('A'))  # décalée de n
    else:
        chiffree = c  # inchangée
    return chiffree


def chiffre_cesar(s, n):
    """Retourne la chaîne s où chaque lettre est décalée de n « vers la droite » (chiffrement de César).

    :param s: la chaîne à chiffrer.
    :type s: str
    :param n: le décalage à appliquer à chaque lettre.
    :returns: la chaîne chiffrée.
    :rtype: str"""
    chiffree = ""  # chaîne chiffrée
    i=0
    while i< len(s):
        chiffree += chiffre_lettre(s[i], n)  
        i+=1
    return chiffree


def dechiffre_cesar(s, n):
    """Retourne la chaîne s (chiffrée avec un décalage de n) déchiffrée, chaque lettre étant décalée en sens inverse.

    :param s: la chaîne chiffrée à déchiffrer.
    :type s: str
    :param n: le décalage utilisé lors du chiffrement.
    :returns: la chaîne déchiffrée.
    :rtype: str"""
    return chiffre_cesar(s,26-n)


message_codé =chiffre_cesar("hello WORLD", 9)
print(message_codé)
message_initial = chiffre_cesar(message_codé,17)
print(message_initial)
````

Définir le module `calculus` contenant les définitions de fonctions suivantes :
- `array_sum` :  cette fonction prend en paramètre un tableau de nombres et retourne la somme des nombres de ce tableau.

````python
def array_sum(tab):
    """Retourne la somme des éléments du tableau tab.

    :param tab: le tableau de nombres dont on calcule la somme.
    :type tab: list
    :returns: la somme des éléments de tab.
    :rtype: list"""
    somme = 0
    i = 0
    while i < len(tab):
        somme += tab[i]
        i += 1
    return somme
````

````python
from math import isclose
def test_array_sum():
    assert array_sum([1, 2, 3]) == 6
    assert array_sum([]) == 0
    assert isclose(array_sum([2.3, 3.4, 4.5]), 10.2)
    print("Test de la fonction array_sum : ok")

test_array_sum()
````

- `array_average` : cette fonction prend en paramètre un tableau de nombres et retourne la moyenne des nombres de ce tableau.

````python
def array_average(tab):
    """Retourne la moyenne des éléments du tableau tab.

    :param tab: le tableau de nombres dont on calcule la moyenne.
    :type tab: list
    :returns: la moyenne des éléments de tab.
    :rtype: list"""
    return array_sum(tab) / len(tab)
````

````python
from math import isclose
def test_array_average():
    assert isclose(array_average([1, 2, 3]), 2)
    assert isclose(array_average([2.3, 3.4, 4.5]), 10.2 / 3)
    assert array_average([]) is None
    print("Test de la fonction array_average : ok")

test_array_average()
````

- `array_variance` : cette fonction prend en paramètre un tableau de nombres et retourne la variance des nombres de ce tableau. La variance des nombres $x_1,\dots, x_n$ est donnée par la formule : $$\frac{1}{n}\sum_{i=1}^n (x_i - \bar x)^2$$ où $\bar x$ correspond à la moyenne des nombres $x_1,\dots, x_n$.

````python
def array_variance(tab):
    """Retourne la variance des éléments du tableau tab.

    :param tab: le tableau de nombres dont on calcule la variance.
    :type tab: list
    :returns: la variance des éléments de tab.
    :rtype: list"""
    m = array_average(tab)
    somme = 0
    i = 0
    while i < len(tab):
        somme += (tab[i] - m)**2
        i += 1
    return somme / len(tab)
````

````python
from math import isclose
def test_array_variance():
    assert isclose(array_variance([2, 2, 2, 2]), 0)
    assert isclose(array_variance([10, 0, 0, 0, 0]), 80/5)
    assert array_variance([]) is None
    print("Test de la fonction array_variance : ok")

test_array_variance()
````
