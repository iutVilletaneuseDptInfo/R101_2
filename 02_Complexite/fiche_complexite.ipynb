### Pourquoi s'intéresser à la complexité ?

Une application qui fonctionne parfaitement avec 10 utilisateurs peut devenir inutilisable avec 100 000 utilisateurs : le code n'a pas changé, mais la taille des données a explosé. La théorie de la complexité permet d'anticiper ce genre de problème, sans attendre que l'application plante en production.


Un algorithme effectue un traitement sur des données. S'il produit le résultat escompté, on dit que l'algorithme est *correct*. On veut de plus que l'algorithme soit *efficace*, c'est-à-dire que le temps de calcul du résultat soit raisonnable.

### Plusieurs chemins mènent à Rome, mais...

**Exemple :** une playlist qui tourne en boucle doit déplacer sa première chanson en fin de liste quand elle vient d'être jouée (*décalage circulaire*) : `[1, 2, 3, 4, 5]` devient `[2, 3, 4, 5, 1]`.

**Première solution :** échanger successivement les éléments consécutifs (1er/2e, puis 2e/3e, etc.) jusqu'au dernier.

**Deuxième solution :** sauvegarder le premier élément, décaler tous les autres d'une case vers la gauche, puis écrire l'élément sauvegardé en dernière position.

Les deux approches sont correctes ; la question posée par ce chapitre est : *lequel est le plus rapide ?*

### Mesure des performances avec `time`

La bibliothèque `time` fournit un chronomètre qui permet de mesurer le temps de calcul.

```python
from time import time

tic = time()  # top départ

###########################################################
#                Traitement à chronométrer                #
###########################################################

tac = time()  # arrêt du chronomètre

print(round(1000 * (tac - tic), 2), " ms") # en ms, arrondi au centième
```

On peut calculer la moyenne d'un certain nombre de mesures des temps d'exécution pour limiter l'impact des fluctuations mesurées (d'autres processus peuvent mobiliser le processeur en parallèle, *etc.*).

### Plusieurs types de complexité

**Complexité linéaire :** un algorithme qui comporte une boucle simple avec un nombre constant $C$ d'opérations élémentaires à chaque itération (pas de boucles imbriquées ni d'appels de fonction) fait au total $Cn$ opérations élémentaires. Le temps de calcul est proportionnel à $n$ : on dit que la complexité est *linéaire*, ou en $O(n)$ (prononcé *grand Ô de n*).

*Note : une opération élémentaire est une « opération simple » arithmétique, logique, d'affectation, etc.*

**Complexité quadratique :** par exemple un algorithme ayant une boucle imbriquée dans laquelle le nombre d'opérations est proportionnel à $n$, elle-même exécutée un nombre de fois proportionnel à $n$ (comme le calcul de toutes les tables de multiplication jusqu'à $n \times n$). La complexité est alors en $O(n^2)$.

### Complexité dans le pire des cas

En général, le nombre d'opérations dépend non seulement de la taille des données, mais des données elles-mêmes (l'*instance*). Par exemple, en recherchant un élément particulier dans un tableau de taille $n$ : avec de la chance, l'élément est trouvé du premier coup (une seule opération) ; mais il est plus utile de prévoir la performance pour un jeu de données défavorable, la **complexité dans le pire des cas**. Pour la recherche, le pire cas se produit quand l'élément n'est pas trouvé (ou trouvé en dernier), ce qui oblige à parcourir le tableau entier : la complexité est alors $O(n)$ dans le pire cas.

### Meilleur cas, pire cas, cas moyen

Le nombre d'opérations dépend aussi des données : **meilleur cas** (données favorables), **pire cas** (données défavorables), **cas moyen** (en moyenne sur les données possibles, sous une hypothèse sur leur distribution : plus délicat à étudier). Sans précision, la complexité d'un algorithme est celle du **pire cas**. Exemple : recherche séquentielle d'un élément, $O(1)$ au mieux, $O(n)$ au pire, $\approx (n+1)/2$ comparaisons en moyenne si l'élément est présent à une position uniformément répartie, donc $O(n)$.

### Récapitulatif des complexités

| Complexité | Nom | Exemples |
|---|---|---|
| $O(1)$ | constante | `tab[i]`, `len(tab)`, `append`, `pop()` |
| $O(\log n)$ | logarithmique | recherche par dichotomie |
| $O(n)$ | linéaire | parcours, maximum, recherche séquentielle, copie, opérations sur les chaînes |
| $O(n \log n)$ | quasi-linéaire | tri fusion |
| $O(n^2)$ | quadratique | boucles imbriquées, tris simples (pire cas), `+=` de chaînes en boucle |
| $O(2^n)$ | exponentielle | `fibonacci` récursif naïf |
