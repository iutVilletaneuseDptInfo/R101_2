## Pourquoi s'intéresser de près aux tableaux ?

Vous développez le tableau de bord d'un tournoi en ligne qui affiche le classement en direct : à chaque score validé, un joueur rejoint la liste ; à chaque abandon ou disqualification, un joueur en sort ; parfois, il faut aussi réorganiser l'ensemble du classement. Toutes ces opérations reposent sur la même structure de données : le **tableau** (la `list` en Python).

Mais toutes les opérations sur un tableau ne se valent pas : certaines sont quasi instantanées, d'autres beaucoup plus coûteuses dès que le tableau grossit. Comprendre ce qui se passe « sous le capot » permet de choisir la bonne opération au bon moment — et d'éviter qu'un classement de 100 000 joueurs ne mette plusieurs secondes à se mettre à jour.

## Qu'est-ce qu'un tableau ?

D'un point de vue algorithmique, un tableau est une structure de données qui permet de stocker une **suite ordonnée** de valeurs $x_0,x_1,…,x_i, …,x_{n−1}$ :
* chaque valeur est repérée par son indice `i` (commençant à 0); 
* l'accès à une case à partir de son indice `i` se fait en **temps constant**, c'est-à-dire que ce temps ne dépend pas de `i`, ni de la longueur `n` du tableau :
  il est donc aussi rapide d'accéder à la première valeur qu'à n'importe quelle autre valeur du tableau. 
  
Ceci est possible car dans un tableau :
* toutes les cases ont la même taille et occupent donc le même espace mémoire ;
* les valeurs sont stockées dans des ***cases contiguës*** de la mémoire de l'ordinateur (appelée RAM : Random Acess Memory) du système informatique.

Cette organisation dans la mémoire a un inconvénient : la taille d'un tableau est fixe.

##  Le type `list`  en Python

Python utilise le type `list`pour implémenter les tableaux. Ce type permet d'adapter **dynamiquement** la taille des tableaux en fonction des éléments insérés. Cependant, il est important de comprendre que ces mécanismes bien qu'automatiques représentent un coût temporel.

Voici les opérations qu'il est possible de faire sur une liste Python.

### Ajout d'une valeur à la fin du tableau

Dans notre classement de tournoi, c'est l'opération la plus fréquente : un joueur termine sa partie, et son score vient s'ajouter à la fin de la liste des scores déjà enregistrés.

En Python, il est possible d'ajouter une valeur `x` à la fin d'une liste `L` avec l'instruction `L.append(x)`.

````python
# Exemple

L = [12, 23, 34, 45]
print(L)

print("Ajout de 56 : ")
L.append(56)
print(L)

print("Ajout de 67 : ")
L.append(67)
print(L)
````

**Complexité :** Cette opération est une opération élémentaire. On considère que sa complexité est $O(1)$. On parle de complexité en *temps constant*.

### Suppression de la dernière valeur du tableau

Si une erreur de saisie vient d'être commise, on peut vouloir annuler le tout dernier score ajouté au classement.

Il est possible de supprimer la dernière valeur du tableau `L` avec l'instruction `L.pop()`. Cette instruction retourne la valeur supprimée.

````python
# Exemple

L = [12, 23, 34, 45]
print(L)

print("Suppression de la dernière valeur : ")
dernier = L.pop()

print("Valeur supprimée :", dernier)

print(L)
````

**Complexité :** Cette opération est une opération élémentaire, sa complexité est $O(1)$. 

### Insertion d'une valeur à un indice donné

Parfois, un joueur doit être inséré à une position précise du classement (par exemple pour respecter un ordre d'arrivée), et pas simplement ajouté à la fin.

Il est possible d'insérer une valeur `x` dans une liste `L` à l'indice `i` avec l'instruction `L.insert(i, x)`.

````python
# Exemple

L = [12, 23, 34, 45]
print(L)

print("Insertion de la valeur 13 à l'indice 1 :")
L.insert(1, 13)
print(L)
````

**Complexité :** Cette opération nécessite de : 
* ajouter une case à la fin du tableau (avec `append`),
* décaler tous les éléments d'une case vers la droite à partir de l'indice `i`,
* insérer la valeur `x` à l'indice `i`.

Dans le pire cas, correspondant à une insertion au début (`i` vaut 0), tous les éléments doivent être décalés d'une case sur la droite. Le décalage est une opération élémentaire donc si la taille du tableau est $n$, alors la complexité dans le pire cas est donc linéaire, c'est-à-dire en $O(n)$.

### Copie d'une liste

Avant de modifier le classement en direct (par exemple pour tester un classement provisoire), on veut souvent garder un instantané de l'état actuel, sans que les modifications ultérieures n'affectent cette sauvegarde.

L'instruction `L.copy()` permet d'obtenir une copie de la liste `L`, c'est-à-dire une deuxième liste en mémoire contenant les mêmes valeurs que `L` dans le même ordre.

````python
# Exemple

L = [12, 23, 34, 45]

copie = L.copy() # copie 

alias = L        # alias
L[0] = 21
print("La liste après modification", L)
print("La copie après modification", copie)
print("L'alias après modification ", alias)
````

**Attention :** L'affectation `alias = L` ne crée pas une copie du tableau, mais un alias ! Les variables `alias` et `L` font référence au même tableau dans la mémoire. Dans notre application de classement, cela voudrait dire que modifier la « sauvegarde » modifierait aussi, par erreur, le classement affiché en direct !

**Complexité :** Cette opération nécessite de :
* créer un tableau vide,
* parcourir les éléments du tableau `L` et les ajouter au fur et à mesure dans le nouveau tableau.

Chaque élément de `L` est ajouté à la fin du nouveau tableau (opération élémentaire). Si `L` contient $n$ éléments, alors la complexité est $O(n)$.

### Tableaux de tableaux : alias, copie, piège de `[[0] * n] * m`

On a vu en partie 1 (chapitres *Tableaux* et *Structures imbriquées*) qu'un tableau peut contenir d'autres tableaux, par exemple pour représenter une grille de `m` lignes et `n` colonnes. On a vu aussi que l'affectation `alias = L` ne crée **pas** une copie : les deux variables désignent le même objet. Ces deux idées se combinent de façon piégeuse.

**Piège 1 : construire une grille avec `*`.** L'expression `[0] * 3` crée bien un tableau de 3 cases indépendantes (des entiers). Mais `[ligne] * 2` ne crée **pas** deux lignes : elle place *deux fois la même ligne* (la même référence) dans le tableau extérieur.

````python
grille = [[0] * 3] * 2    # on veut 2 lignes de 3 zéros
print(grille)
grille[0][0] = 1          # on ne modifie qu'une seule case...
print(grille)             # ...mais les deux lignes ont changé !
print(grille[0] is grille[1])   # True : les deux lignes sont le même tableau
````

Pour obtenir des lignes **indépendantes**, il faut créer chaque ligne séparément, par exemple avec une boucle et `append` :

````python
grille = []
i = 0
while i < 2:
    grille.append([0] * 3)   # une nouvelle ligne à chaque tour
    i += 1
grille[0][0] = 1
print(grille)                # seule la première ligne a changé
````

**Piège 2 : la copie d'un tableau de tableaux.** `L.copy()` est une copie **superficielle** : elle recopie les cases du tableau extérieur, mais ces cases contiennent des *références* vers les mêmes lignes. La copie et l'original partagent donc leurs lignes. Pour obtenir une copie indépendante (copie **profonde**), il faut aussi copier chaque ligne.

````python
grille = [[1, 2], [3, 4]]
copie = grille.copy()
copie[0][0] = 99
print(grille)            # [[99, 2], [3, 4]] : l'original a été modifié !

grille = [[1, 2], [3, 4]]
copie = []
i = 0
while i < len(grille):
    copie.append(grille[i].copy())   # on copie chaque ligne
    i += 1
copie[0][0] = 99
print(grille)               # [[1, 2], [3, 4]] : l'original est intact
````

*(Python propose aussi la fonction `deepcopy` du module `copy` qui réalise cette copie profonde automatiquement, quelle que soit la profondeur d'imbrication.)*

**Complexité :** copier une grille de `m` lignes de `n` colonnes coûte $O(m \times n)$ en copie profonde, contre $O(m)$ seulement en copie superficielle.

### Autres opérations sur les listes Python

* `L.reverse()`: inverse les items d'une liste. 
* `L.remove(x)` : supprime la première occurence de la valeur `x` (**attention :** produit une erreur si `x` n'est pas dans la liste).
* `L.count(x)` : compte le nombre d'occurences d'une valeur `x`.
* `L.index(x)` : permet de connaître la première position de la valeur cherchée `x` (**attention :** produit une erreur si `x` n'est pas dans la liste).
* `L.sort()` : trie la liste `L`.
* `x in L` : vaut `True` si `x` est un élément de `L`, `False` sinon.

````python
# Exemple

L = [1, -2, 0, 3, 0, -4, 0, 5]
print(L)
print("Inversion de la liste")
L.reverse()
print(L)

print("Nombre de 0 dans la liste :", L.count(0))

val = -2
if val in L:
    print("La première occurence de", val, "se trouve à l'indice",
          L.index(val))
    print("On supprime cette première occurence !")
    L.remove(val)
    print(L)
else:
    print(val, "n'appartient pas à la liste")

print("Tri de la liste")
L.sort()
print(L)
````

### Complexité de l'insertion et de la suppression en fin de tableau

Revenons à notre classement de tournoi : si des milliers de scores sont ajoutés les uns après les autres, l'ajout en fin de tableau reste-t-il vraiment instantané ? La réponse est plus subtile qu'il n'y paraît.

Dans ce cours, l'insertion en fin de tableau est considérée comme une opération élémentaire. Ceci n'est pas complètement vrai. En effet, pour ajouter une case à la fin du tableau, il faut qu'il y ait de la place dans la RAM juste à côté du tableau puisque les valeurs sont contiguës. Si ce n'est pas le cas, il faut auparavant "déplacer" le tableau dans un autre endroit de la mémoire, c'est-à-dire créer une copie du tableau dans un endroit de la mémoire où il y a suffisamment de place et supprimer l'original. Cette opération de copie a une complexité linéaire, ce qui implique qu'ajouter une case à la fin d'un tableau a une complexité linéaire dans le pire cas.

Pour éviter d'effectuer trop de copies lors d'insertions en fin de tableau, l'interpréteur réserve plus de place que nécessaire pour créer un tableau. Les premières insertions sont faites en temps constant puisqu'il y a de la place disponible. S'il faut effectuer une copie du tableau, l'interpréteur double la réservation de mémoire et les insertions suivantes seront en temps constant. De cette manière, il y a peu de copies en moyenne et l'on peut considérer l'insertion en fin de tableau comme une opération élémentaire (on parle de complexité amortie constante).

Le code suivant montre l'évolution du temps moyen d'insertion en fin de tableau en fonction de la taille du tableau.

````python
from time import time


def mesure_insertion(n, nb_iter=100):
    """Mesure le temps d'insertion d'un élément en fin de tableau, 
       en fonction de la taille du tableau.

    :param n: la taille maximale de tableau testée.
    :type n: int
    :param nb_iter: le nombre de mesures moyennées pour chaque taille.
    :type nb_iter: int
    :returns: un tableau de taille n où la i-ème case indique le temps moyen (en ms) pour ajouter un élément à un tableau de taille i.
    :rtype: list"""
    tps = []
    i = 0
    while i < n:
        tps.append(0.)
        i += 1

    j = 0
    while j < nb_iter:
        t = []
        i = 0
        while i < n:
            tic = time()
            t.append(1)
            tac=time()
            tps[i] += tac - tic
            i += 1
        j += 1

    i = 0
    while i < n:
        tps[i] *= 100 / nb_iter
        i += 1
    return tps
````

````python
def test_mesure_insertion():
    resultat = mesure_insertion(5, 10)
    assert len(resultat) == 5
    assert all(t >= 0 for t in resultat)
    print("Test de la fonction mesure_insertion : ok")

test_mesure_insertion()
````

### Le type `array` du module `numpy`

Le type `list` peut contenir différents types. Par exemple, `t = [10,8.6,"coucou"]`. De plus, les listes peuvent adapter leur capacité en fonction des besoins. Cette polyvalence se fait au prix d'une perte de performance des algorithmes. 

Le type `array` du module [NumPy](https://numpy.org/) (auquel l'instruction `import numpy` permet d'accéder) est 
- "plus contraint" : seul le stockage d'éléments de même type dans des tableaux éventuellement multidimensionnels est possible, 
- et plus efficace : cette restriction va de paire avec des algorithmes  plus performants. 

Le type `array` est particulièrement bien adapté au calcul scientifique pour des données de grande taille.
