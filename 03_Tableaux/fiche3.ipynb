### Pourquoi s'intéresser de près aux tableaux ?

Un tableau (`list` en Python) permet de stocker une suite ordonnée de valeurs, comme un classement de tournoi en direct. Mais toutes les opérations sur un tableau ne se valent pas : certaines sont quasi instantanées, d'autres coûtent cher dès que le tableau grossit. Savoir laquelle utiliser au bon moment fait toute la différence sur un tableau de grande taille.

### Qu'est-ce qu'un tableau ?

D'un point de vue algorithmique, un tableau est une structure de données qui permet de stocker une suite ordonnée de valeurs $x_0, x_1, \dots, x_i, \dots, x_{n-1}$ :

* chaque valeur est repérée par son indice $i$ (commençant à 0),
* l'accès à une case à partir de son indice $i$ se fait en temps constant : ce temps ne dépend ni de $i$, ni de la longueur $n$ du tableau (aussi rapide d'accéder à la première valeur qu'à la 200ᵉ).

Ceci est possible car dans un tableau, toutes les cases ont la même taille et sont stockées dans des cases contiguës de la mémoire de l'ordinateur (la RAM). Le nombre de cases d'un tableau non dynamique est fixe.

### Le type `list` en Python

Python utilise le type `list` pour implémenter les tableaux. Le langage permet aux listes d'adapter dynamiquement leur taille en fonction des insertions/suppressions ; ces mécanismes, bien qu'automatiques, ont un coût temporel.

#### Opérations de base sur les listes Python et complexité

(`L` est une variable de type `list` de taille $n$)

Opération | Syntaxe | Complexité
-----------|----------|-------------
Création d'une liste vide | `[]` | $O(1)$
Lecture d'un élément | `x = L[i]` | $O(1)$
Écriture d'un élément | `L[i] = x` | $O(1)$
Connaître la taille de la liste | `len(L)` | $O(1)$
Ajout d'un élément à la fin | `L.append(x)` | $O(1)$
Suppression du dernier élément | `L.pop()` | $O(1)$
Insertion d'un élément à l'indice `i` | `L.insert(i, x)` | $O(n)$
Suppression de l'élément d'indice `i` | `L.pop(i)` | $O(n)$
Copie d'une liste | `L2 = L.copy()` | $O(n)$

#### Exemple d'utilisation des opérations de base

```python
L = [12, 23, 34, 45]      # L vaut [12, 23, 34, 45]
valeur = L.pop()          # L vaut [12, 23, 34] et valeur vaut 45
L.append(56)              # L vaut [12, 23, 34, 56]
L.append(67)              # L vaut [12, 23, 34, 56, 67]
L.insert(1, 13)           # L vaut [12, 13, 23, 34, 56, 67]
valeur = L.pop(2)         # L vaut [12, 13, 34, 56, 67] et valeur vaut 23
L2 = L.copy()             # L vaut [12, 13, 34, 56, 67] et
                          # L2 est un autre tableau contenant [12, 13, 34, 56, 67]
print(L, valeur, L2)
```

#### Autres opérations sur les listes Python

* `L.reverse()` : inverse les éléments d'une liste.
* `L.remove(x)` : supprime la première occurrence de `x` (**attention :** génère une erreur si `x` n'est pas dans la liste).
* `L.count(x)` : compte le nombre d'occurrences de la valeur `x`.
* `L.index(x)` : donne la position de `x` (**attention :** génère une erreur si `x` n'est pas dans la liste).
* `L.sort()` : trie la liste `L`.
* `x in L` : vaut `True` si `x` est un élément de `L`, `False` sinon.

```python
L = [1, -2, 0, 3, 0, -2, 0, 5]
L.reverse()          # L vaut [5, 0, -2, 0, 3, 0, -2, 1]
nb0 = L.count(0)     # nb0 vaut 3
val = -2
if val in L:         # vaut True car -2 est dans la liste
    indice = L.index(val)  # indice vaut 2
    L.remove(val)          # L vaut [5, 0, 0, 3, 0, -2, 1]
L.sort()             # L vaut [-2, 0, 0, 0, 1, 3, 5]
print(L, nb0, indice)
```

### Tableaux de tableaux : alias et copie

`[[0] * 3] * 2` ne crée **pas** deux lignes indépendantes : la même ligne est présente deux fois, modifier l'une modifie l'autre. Pour une grille indépendante, créer chaque ligne séparément (boucle + `append`).

`L.copy()` est une copie **superficielle** : pour un tableau de tableaux, les lignes restent partagées avec l'original. Pour une copie **profonde**, copier aussi chaque ligne (complexité $O(m \times n)$ pour `m` lignes de `n` colonnes) ou utiliser `deepcopy` du module `copy`.
